```python
# =====================================================
# Mobile Game In-App Purchases Analysis (Excel from GitHub)
# =====================================================

# ------------------------------
# Imports
# ------------------------------
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set(style="whitegrid")

# ------------------------------
# Step 1: Load Dataset from GitHub
# ------------------------------
RAW_URL = "https://raw.githubusercontent.com/shra1802/Kaggle-Dataset-Projects/Mobile-Game-In-App-Purchases-Dataset-2025/mobile_game_inapp_purchases.xlsx"

df = pd.read_excel(RAW_URL)

print("✅ Loaded dataset from GitHub!")
print("Shape:", df.shape)
df.head()
```

```python
# ------------------------------
# Data Cleaning
# ------------------------------
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Gender'] = df['Gender'].fillna('Unknown')
df['Device'] = df['Device'].fillna(df['Device'].mode()[0])
df['InAppPurchaseAmount'] = df['InAppPurchaseAmount'].fillna(0)

categorical_cols = ['Country', 'GameGenre', 'SpendingSegment', 'PaymentMethod']
df[categorical_cols] = df[categorical_cols].fillna('Unknown')

print("\nMissing values after cleaning:")
df.isnull().sum()
```

**Exploratory Data Analysis**

```python
# Log transformation for skew
df['LogPurchase'] = np.log1p(df['InAppPurchaseAmount'])

plt.figure(figsize=(8,5))
sns.histplot(df['LogPurchase'], kde=True, bins=50)
plt.title("Distribution of Log-Transformed In-App Purchase Amount")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='SpendingSegment', data=df,
              order=['Whale','Dolphin','Minnow'])
plt.title("Spending Segment Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Device', y='InAppPurchaseAmount', data=df)
plt.title("Revenue by Device")
plt.show()
```

**A/B Testing (iOS vs Android)**

```python
group_ios = df[df['Device']=="iOS"]['InAppPurchaseAmount']
group_android = df[df['Device']=="Android"]['InAppPurchaseAmount']

plt.figure(figsize=(6,4))
sns.boxplot(x='Device', y='InAppPurchaseAmount',
            data=df[df['Device'].isin(['iOS','Android'])])
plt.title("iOS vs Android In-App Revenue")
plt.show()
```

```python
ios_sample = group_ios.sample(min(500, len(group_ios)), random_state=42)
android_sample = group_android.sample(min(500, len(group_android)), random_state=42)

print("Shapiro Normality Tests:")
print("iOS:", stats.shapiro(ios_sample))
print("Android:", stats.shapiro(android_sample))
```

```python
if stats.shapiro(ios_sample).pvalue > 0.05 and stats.shapiro(android_sample).pvalue > 0.05:
    stat, p_value = stats.ttest_ind(group_ios, group_android, equal_var=False)
    test = "Independent T-Test"
else:
    stat, p_value = stats.mannwhitneyu(group_ios, group_android,
                                        alternative='two-sided')
    test = "Mann-Whitney U Test"

print(f"\n{test} Results")
print("Statistic:", stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("✅ Statistically significant difference between iOS and Android.")
else:
    print("❌ No statistically significant difference.")
```

**Business Insights**

```python
print("\nAvg Revenue by Game Genre:")
print(df.groupby('GameGenre')['InAppPurchaseAmount'].mean().sort_values(ascending=False))
```

```python
print("\nAvg Revenue by Spending Segment:")
print(df.groupby('SpendingSegment')['InAppPurchaseAmount'].mean())
```

```python
df['AgeGroup'] = pd.cut(df['Age'],
                        bins=[0,18,25,35,50,100],
                        labels=['<18','18-25','26-35','36-50','50+'])

print("\nAvg Revenue by Age Group:")
print(df.groupby('AgeGroup')['InAppPurchaseAmount'].mean())
```

```python
pivot = df.pivot_table(values='InAppPurchaseAmount',
                       index='Country', columns='Device',
                       aggfunc='mean')

plt.figure(figsize=(10,6))
sns.heatmap(pivot, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Avg Revenue by Country & Device")
plt.show()
```
